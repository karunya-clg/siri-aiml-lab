```python

p_burglary = 0.002
p_earthquake = 0.001

p_alarm_given_burglary_and_earthquake = 0.94
p_alarm_given_burglary_and_no_earthquake = 0.95
p_alarm_given_no_burglary_and_earthquake = 0.31
p_alarm_given_no_burglary_and_no_earthquake = 0.001

p_david_calls_given_alarm = 0.91
p_david_does_not_call_given_alarm = 0.09
p_david_calls_given_no_alarm = 0.05
p_david_does_not_call_given_no_alarm = 0.95

p_sophia_calls_given_alarm = 0.75
p_sophia_does_not_call_given_alarm = 0.25
p_sophia_calls_given_no_alarm = 0.02
p_sophia_does_not_call_given_no_alarm = 0.98


def joint_probability(A, B, E, D, S):
    if B and E:
        p = p_alarm_given_burglary_and_earthquake
    elif B:
        p = p_alarm_given_burglary_and_no_earthquake
    elif E:
        p = p_alarm_given_no_burglary_and_earthquake
    else:
        p = p_alarm_given_no_burglary_and_no_earthquake

    p_alarm = p if A else 1-p

    d = (p_david_calls_given_alarm if D else p_david_does_not_call_given_alarm) if A else (p_david_calls_given_no_alarm if D else p_david_does_not_call_given_no_alarm)

    s = (p_sophia_calls_given_alarm if S else p_sophia_does_not_call_given_alarm) if A else (p_sophia_calls_given_no_alarm if S else p_sophia_does_not_call_given_no_alarm)

    b = p_burglary if B else 1-p_burglary
    e = p_earthquake if E else 1-p_earthquake

    return b * e * p_alarm * d * s


print("Output:")
r = joint_probability(True, False, True, True, False)
print(f"Joint Probability (A, not B, E, D, not S): {r:.10f}")

print("\nTest Case 1: A=True, B=False, E=False, D=True, S=True")
r = joint_probability(True, False, False, True, True)
print(f"Joint Probability (A, not B, not E, D, S): {r:.10f}")

print("\nTest Case 2: A=True, B=True, E=False, D=True, S=True")
r = joint_probability(True, True, False, True, True)
print(f"Joint Probability (A, B, not E, D, S): {r:.10f}")

print("\nTest Case 3: A=True, B=False, E=True, D=True, S=False")
r = joint_probability(True, False, True, True, False)
print(f"Joint Probability (A, not B, E, D, not S): {r:.10f}")
```
