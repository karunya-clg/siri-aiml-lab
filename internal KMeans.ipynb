```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.cluster import KMeans

iris=datasets.load_iris()
df=pd.DataFrame(iris.data)
df.columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
plt.figure(figsize=(15, 10))

model1=KMeans(n_clusters=2, random_state=0)
model1.fit(df)
plt.subplot(2,3,1)
plt.scatter(df['petal_length'], df['petal_width'], c=model1.labels_)
print("TC1: 2 Clusters")

model2=KMeans(n_clusters=3, random_state=0)
plt.subplot(2,3,2)
model2.fit(df)
plt.scatter(df['petal_length'], df['petal_width'], c=model2.labels_)
print("TC2: 3 Clusters")

model3=KMeans(n_clusters=3, max_iter=300, random_state=0)
model3.fit(df)
plt.subplot(2,3,3)
plt.scatter(df['petal_length'], df['petal_width'], c=model3.labels_)
print("TC3: convergence")

print("chek convergence")
print("No of iterations: ", model3.n_iter_)
print("cluster centroids", model3.cluster_centers_)

plt.show()
```
